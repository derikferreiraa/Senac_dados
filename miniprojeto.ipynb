# Gasolina no Rio de Janeiro · 2024 e 2025

Entre os municípios fluminenses pesquisados pela ANP, quais tiveram o maior preço médio de venda da gasolina comum no período?

Coloque os quatro CSVs semestrais de combustíveis automotivos na mesma pasta deste notebook e execute as células em ordem.

## 1. Ler os quatro semestres

Cada linha da pesquisa representa uma observação de preço. Vamos começar com uma tabela para cada semestre.

```python
import pandas as pd
base_2024_1 = pd.read_csv("Preços semestrais - AUTOMOTIVOS_2024.01.csv", sep=";", decimal=",")
base_2024_2 = pd.read_csv("Preços semestrais - AUTOMOTIVOS_2024.02.csv", sep=";", decimal=",")
base_2025_1 = pd.read_csv("Preços semestrais - AUTOMOTIVOS_2025.01.csv", sep=";", decimal=",")
base_2025_2 = pd.read_csv("Preços semestrais - AUTOMOTIVOS_2025.02.csv", sep=";", decimal=",")
```

## 2. Reunir as coletas

Os arquivos têm as mesmas colunas e cobrem períodos diferentes. Empilhamos suas linhas e comparamos o total obtido com a soma das quatro bases.

```python
bases = [base_2024_1, base_2024_2, base_2025_1, base_2025_2]
precos = pd.concat(bases, ignore_index=True)
total_esperado = len(base_2024_1) + len(base_2024_2) + len(base_2025_1) + len(base_2025_2)
print("Linhas antes e depois da união:", total_esperado, len(precos))
precos.head()
```

## 3. Selecionar a gasolina no RJ

O campo de estado permite selecionar o Rio de Janeiro inteiro. O rótulo GASOLINA é o mesmo usado no exercício de aula.

```python
no_rj = precos["Estado - Sigla"] == "RJ"
gasolina_comum = precos["Produto"] == "GASOLINA"
gasolina_rj = precos.loc[no_rj & gasolina_comum, ["Municipio", "Valor de Venda"]]
print("Coletas selecionadas:", len(gasolina_rj))
gasolina_rj.head()
```

## 4. Comparar os municípios

Calculamos o preço médio observado em cada município nos quatro semestres juntos. A primeira linha do resultado indica o maior valor médio do recorte.

```python
media_municipal = gasolina_rj.groupby("Municipio")["Valor de Venda"].mean()
ranking = media_municipal.sort_values(ascending=False)
ranking.head(10).round(2)
```

### Como ler o resultado

Os valores estão em reais por litro e descrevem as coletas do período. Um posto pesquisado em mais semanas contribui com mais observações. O ranking compara os preços observados e não mede o volume vendido. Para registrar a conclusão, execute o notebook e anote o município que aparece no topo e seu preço médio.
